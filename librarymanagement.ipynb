```python
taken_books = []

class Book:
    def __init__(self, book_id, name, author, quantity):
        self.book_id = book_id
        self.name = name
        self.author = author
        self.quantity = quantity

    def add_book(self):
        choice = int(input("Enter the number of books you want to add: "))
        for i in range(choice):
            book_name = input("Enter the book: ")
            book_id = int(input("Enter the id of the book: "))
            author = input("Enter the author of the book: ")
            quantity = int(input("Enter the quantity of the book: "))
            b = Book(book_id, book_name, author, quantity)
            Books.append(b)

    def display(self):
        if not Books:
            print("No books available. Please add books.")
        else:
            print("Displaying records")
            print("-" * 40)
            for book in Books:
                print(f"Book id: {book.book_id}")
                print(f"Book Name: {book.name}")
                print(f"Author: {book.author}")
                print(f"Quantity: {book.quantity}")
                print("-" * 40)

    def linear_search(self):
        search_item = int(input("Enter the book Id: "))
        for book in Books:
            if book.book_id == search_item:
                print("Book found!")
                print("Book id:", book.book_id)
                print("Title:", book.name)
                print("Author:", book.author)
                print("Quantity:", book.quantity)
                return
        print("Book not found.")

    def issue(self):
        book_id = int(input("Enter the Book ID: "))
        quantity = int(input("Enter the quantity: "))
        for book in Books:
            if book.book_id == book_id:
                if book.quantity == 0:
                    print("You cannot issue this book.")
                elif quantity > book.quantity:
                    print("Not enough books available.")
                else:
                    book.quantity -= quantity
                    t = (book.book_id, book.name, book.author, quantity)
                    taken_books.append(t)
                    print("Book issued successfully!")
                    print("Remaining quantity:", book.quantity)
                return
        print("Book not found.")

    def return_book(self):
        book_id = int(input("Enter the Book ID: "))
        quantity = int(input("Enter the quantity: "))
    
        for book in Books:
            if book.book_id == book_id:
                book.quantity += quantity
    
                for i in range(len(taken_books)):
                    if taken_books[i][0] == book_id:
                        old_quantity = taken_books[i][3]
    
                        if quantity <= old_quantity:
                            new_quantity = old_quantity - quantity
    
                            if new_quantity == 0:
                                taken_books.pop(i)
                            else:
                                taken_books[i] = (
                                    taken_books[i][0],
                                    taken_books[i][1],
                                    taken_books[i][2],
                                    new_quantity
                                )
    
                            print("Book returned successfully!")
                            print("Available quantity:", book.quantity)
                            return

            print("No issued record found.")
            return

    print("Book not found.")

    def delete_book(self):
        delete = input("Enter the book name: ")
        for book in Books:
            if book.name.lower() == delete.lower():
                Books.remove(book)
                print("Deleted book successfully!")
                return
        print("Book not found.")

    def delete_all(self):
        Books.clear()
        print("Your book list has been cleared")

    def update_book(self):
        updatebook = input("Enter the book name you want to update: ")
        for book in Books:
            if book.name.lower() == updatebook.lower():
                print("Book found!")
                print(f"ID: {book.book_id}, Name: {book.name}, Author: {book.author}, Quantity: {book.quantity}")
                book.book_id = int(input("Enter new Book ID: "))
                book.name = input("Enter new Book Name: ")
                book.author = input("Enter new Author: ")
                book.quantity = int(input("Enter new Quantity: "))
                print("Book updated successfully!")
                return
        print("Book not found.")

    def display_taken_books(self):
        if not taken_books:
            print("No books have been issued.")
        else:
            print("Issued Books:")
            print("-" * 40)
            for t in taken_books:
                print(f"Book ID: {t[0]}, Name: {t[1]}, Author: {t[2]}, Quantity Issued: {t[3]}")
                print("-" * 40)

    def save_taken_books(self):
        with open("takenbooks.txt", "w") as f:
            for t in taken_books:
                f.write(
                    f"Book ID: {t[0]}, Name: {t[1]}, "
                    f"Author: {t[2]}, Quantity Issued: {t[3]}\n"
                )

Books = [
    Book(101, "Python Programming", "Eric Matthes", 5),
    Book(102, "Clean Code", "Robert Martin", 3),
    Book(103, "The Pragmatic Programmer", "Andrew Hunt", 4),
    Book(104, "Introduction to Algorithms", "Thomas Cormen", 2),
    Book(105, "Data Structures", "Narasimha Karumanchi", 6),
    Book(106, "Machine Learning", "Tom Mitchell", 3),
    Book(107, "Computer Networks", "Andrew Tanenbaum", 4),
    Book(108, "Operating System Concepts", "Silberschatz", 5),
    Book(109, "Database System Concepts", "Silberschatz", 2),
    Book(110, "Artificial Intelligence", "Stuart Russell", 3)
]

b = Book(0, "", "", 0)  # dummy object

while True:
    print("\n===== Library Menu =====")
    print("1. Add Book")
    print("2. Display Books")
    print("3. Search Book")
    print("4. Issue Book")
    print("5. Return Book")
    print("6. Delete One Book")
    print("7. Delete All Books")
    print("8. Update Book")
    print("9. Display Issued Books")
    print("10. Save Issued Books")
    print("11. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        b.add_book()
    elif choice == 2:
        b.display()
    elif choice == 3:
        b.linear_search()
    elif choice == 4:
        b.issue()
    elif choice == 5:
        b.return_book()
    elif choice == 6:
        b.delete_book()
    elif choice == 7:
        b.delete_all()
    elif choice == 8:
        b.update_book()
    elif choice == 9:
        b.display_taken_books()
    elif choice == 10:
        b.save_taken_books()
    elif choice == 11:
        print("Exiting program... Goodbye!")
        break
    else:
        print("Invalid choice. Please try again.")
```
